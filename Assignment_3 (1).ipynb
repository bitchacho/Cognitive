```python
#Q4

from sklearn.datasets import load_iris

iris_data = load_iris(as_frame=True)
iris_df = iris_data.frame
iris_df["species"] = iris_df["target"].map(dict(enumerate(iris_data.target_names)))
iris_df = iris_df.drop(columns=["target"])
iris_df.to_csv("iris.csv", index=False)

iris_csv = pd.read_csv("iris.csv")
print(iris_csv.head())
```

```python
#Q5

iris_modified = iris_csv.drop(index=4)
dropped_column = iris_modified.columns[3]
iris_modified = iris_modified.drop(columns=[dropped_column])

print("Dropped column:", dropped_column)
print("Shape after deletion:", iris_modified.shape)
print(iris_modified.head())
```

```python
#Q6

employees_data = {
    "Employee_ID": [101, 102, 103, 104, 105],
    "Name": ["Alice", "Bob", "Charlie", "Diana", "Edward"],
    "Department": ["HR", "IT", "IT", "Marketing", "Sales"],
    "Age": [29, 34, 41, 28, 38],
    "Salary": [50000, 70000, 65000, 55000, 60000],
    "Years_of_Experience": [4, 8, 10, 3, 12],
    "Joining_Date": ["2020-03-15", "2017-07-19", "2013-06-01", "2021-02-10", "2010-11-25"],
    "Gender": ["Female", "Male", "Male", "Female", "Male"],
    "Bonus": [5000, 7000, 6000, 4500, 5000],
    "Rating": [4.5, 4.0, 3.8, 4.7, 3.5]
}
employees_df = pd.DataFrame(employees_data)
employees_df.to_csv("employees.csv", index=False)
employees_df = pd.read_csv("employees.csv")
print(employees_df)

#a) Shape of the DataFrame
print("\na) Shape:", employees_df.shape)

#b) Summary - data types and non-null counts
print("\nb) Info:")
employees_df.info()

#c) Descriptive statistics
print("\nc) Describe:")
print(employees_df.describe())

#d) First 5 rows and last 3 rows
print("\nd) Head:")
print(employees_df.head())
print("\nd) Tail:")
print(employees_df.tail(3))

#e) Statistics
print("\ne) i. Average salary:", employees_df["Salary"].mean())
print("e) ii. Total bonus:", employees_df["Bonus"].sum())
print("e) iii. Youngest employee age:", employees_df["Age"].min())
print("e) iv. Highest rating:", employees_df["Rating"].max())

#f) Sort by Salary descending
print("\nf) Sorted by Salary (desc):")
print(employees_df.sort_values("Salary", ascending=False))

#g) Performance category column
def categorize(rating):
    if rating >= 4.5:
        return "Excellent"
    elif rating >= 4.0:
        return "Good"
    else:
        return "Average"

employees_df["Performance_Category"] = employees_df["Rating"].apply(categorize)
print("\ng) Performance category:")
print(employees_df[["Name", "Rating", "Performance_Category"]])

#h) Missing values
print("\nh) Missing values:")
print(employees_df.isnull().sum())

#i) Rename Employee_ID to ID
employees_df = employees_df.rename(columns={"Employee_ID": "ID"})
print("\ni) Columns after rename:", employees_df.columns.tolist())

#j) Filtering
print("\nj) i. More than 5 years experience:")
print(employees_df[employees_df["Years_of_Experience"] > 5])
print("\nj) ii. IT department:")
print(employees_df[employees_df["Department"] == "IT"])

#k) Tax column
employees_df["Tax"] = employees_df["Salary"] * 0.10
print("\nk) Tax column added:")
print(employees_df[["Name", "Salary", "Tax"]])

#l) Save modified DataFrame
employees_df.to_csv("employees_modified.csv", index=False)
print("\nl) Saved modified dataset to employees_modified.csv")
```
