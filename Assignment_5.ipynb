```python
#Q1

import numpy as np

arr = np.array([1, 2, 3, 4, 5])
print("Original array:", arr)

print("a) Add 2 to all elements:", arr + 2)
print("b) Multiply all elements by 3:", arr * 3)
print("c) Divide every element by 2:", arr / 2)
```

```python
#Q2a

arr = np.array([1, 2, 3, 6, 4, 5])
print("Original array:", arr)
print("Reversed array:", arr[::-1])
```

```python
#Q2b

x = np.array([1, 2, 3, 4, 5, 1, 2, 1, 1, 1])
values, counts = np.unique(x, return_counts=True)
most_freq_x = values[np.argmax(counts)]
indices_x = np.where(x == most_freq_x)[0]
print("x:", x)
print("Most frequent value:", most_freq_x)
print("Indices:", indices_x)

y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3])
values, counts = np.unique(y, return_counts=True)
most_freq_y = values[np.argmax(counts)]
indices_y = np.where(y == most_freq_y)[0]
print("\ny:", y)
print("Most frequent value:", most_freq_y)
print("Indices:", indices_y)
```

```python
#Q3

arr = np.array([[10, 20, 30], [40, 50, 60], [70, 80, 90]])
print("Array:")
print(arr)

print("a) 1st row, 2nd column:", arr[0, 1])
print("b) 3rd row, 1st column:", arr[2, 0])
```

```python
#Q4

Garv = np.linspace(10, 100, 25)
print("Garv:", Garv)

print("\nDimensions:", Garv.ndim)
print("Shape:", Garv.shape)
print("Total elements:", Garv.size)
print("Data type:", Garv.dtype)
print("Total bytes consumed:", Garv.nbytes)

Garv_transpose_reshape = Garv.reshape(25, 1)
print("\nTranspose using reshape(25,1):")
print(Garv_transpose_reshape)

Garv_T = Garv.T
print("\nUsing .T attribute:")
print(Garv_T)
print("Shape after .T:", Garv_T.shape)
print("\nNote: .T has no effect here since a 1-D array has only one axis to")
print("transpose against, so .T leaves it unchanged (still shape (25,)).")
print("reshape(25,1) is what actually converts it into a column vector.")
```

```python
#Q5

ucs420_Garv = np.array([[10, 20, 30, 40],
                         [50, 60, 70, 80],
                         [90, 15, 20, 35]])
print("ucs420_Garv:")
print(ucs420_Garv)

print("\nMean:", np.mean(ucs420_Garv))
print("Median:", np.median(ucs420_Garv))
print("Max:", np.max(ucs420_Garv))
print("Min:", np.min(ucs420_Garv))
print("Unique elements:", np.unique(ucs420_Garv))

reshaped_ucs420_Garv = ucs420_Garv.reshape(4, 3)
print("\nreshaped_ucs420_Garv (4x3):")
print(reshaped_ucs420_Garv)

resized_ucs420_Garv = np.resize(ucs420_Garv, (2, 3))
print("\nresized_ucs420_Garv (2x3):")
print(resized_ucs420_Garv)
```
