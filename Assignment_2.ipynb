```python
#Q1

roll_no=int(input("Enter your roll number:"))
digits=[int(x) for x in str(roll_no)]
l=[x*10 for x in digits]
print("l:",l)

l.append(100)
print("After append:",l)

l.insert(4,200)
print("After insert:",l)

l.remove(100)
print("After remove:",l)

l.pop()
print("After pop:",l)

l.sort()
print("Ascending:",l)

l.sort(reverse=True)
print("Descending:", l)

print("First 3 elements:",l[:3])
print("Last 3 elements:",l[-3:])

avg=sum(l)/len(l)

new_list = [x for x in l if x>avg]

print("Average:",avg)
print("Elements greater than average:",new_list)
```

```python
#Q2

scores = tuple(l[:8])
print("Scores:",scores)

highest=max(scores)
highest_index=scores.index(highest)
lowest = min(scores)
lowest_count = scores.count(lowest)

print("Highest score:",highest)
print("Highest index:",highest_index)
print("Lowest score:",lowest)
print("Lowest count:",lowest_count)

reversed_tuple=tuple(reversed(scores))
print("Reversed tuple:",reversed_tuple)

score = int(input("Enter a score:"))
if score in scores:
    print("First occurrence index:",scores.index(score))
else:
    print("Not present")

try:
    scores[0]=100
except TypeError as e:
    print(e)

first,second,*remaining=scores
print("First score:",first)
print("Second score:",second)
print("Remaining scores:",remaining)
```

```python
#Q3

import random

roll_no=int(input("Enter your roll number: "))
random.seed(roll_no)
numbers=[random.randint(100, 900) for i in range(100)]
print("Numbers:",numbers)

odd=[x for x in numbers if x%2!=0]
print("Odd numbers:",odd)
print("Odd count:",len(odd))

even=[x for x in numbers if x%2==0]
print("Even numbers:",even)
print("Even count:",len(even))

def is_prime(n):
    if n<2:
        return False
        
    for i in range(2,int(n**0.5)+1):
        if n%i==0:
            return False
    return True

prime=[x for x in numbers if is_prime(x)]
print("Prime numbers:",prime)
print("Prime count:",len(prime))

count = {}
for x in numbers:
    if x in count:
        count[x]=count[x]+1
    else:
        count[x] =1

most_number=max(count,key=count.get)
print("Most frequent number =", most_number)
print("Number of times =", count[most_number])
```

```python
#Q4

roll_no=int(input("Enter your roll number:"))
digits=[int(x) for x in str(roll_no)]

A={x*7 for x in digits}
B={x*9 for x in digits}
print("A:",A)
print("B:",B)

print("Union:",A | B)
print("Intersection:",A & B)

print("A-B:",A-B)
print("B-A:",B-A)

print("Symmetric difference:",A^B)

print("A is subset of B:",A.issubset(B))
print("B is superset of A:",B.issuperset(A))

X=int(input("Enter a value:"))

A.discard(X)
print("A after discard =", A)
```
