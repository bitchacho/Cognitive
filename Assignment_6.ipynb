```python
#Q1

import numpy as np
import matplotlib.pyplot as plt

M = float(input("Enter a value for M: "))

x = np.linspace(-10, 10, 200)
y1 = M * x**2
y2 = M * np.sin(x)

fig, ax1 = plt.subplots(figsize=(8, 5))

line1, = ax1.plot(x, y1, color="blue", linestyle="-", label="y = M*x^2")
ax1.set_xlabel("x")
ax1.set_ylabel("y = M*x^2", color="blue")
ax1.tick_params(axis="y", labelcolor="blue")

ax2 = ax1.twinx()
line2, = ax2.plot(x, y2, color="red", linestyle="--", label="y = M*sin(x)")
ax2.set_ylabel("y = M*sin(x)", color="red")
ax2.tick_params(axis="y", labelcolor="red")

ax1.set_title(f"Plot of y = M*x^2 and y = M*sin(x) for M = {M}")
ax1.legend(handles=[line1, line2], loc="upper center")
ax1.grid(True)

plt.show()
```

```python
#Q2

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

subjects = ["Maths", "Physics", "Chemistry", "English", "Computer Science"]
scores = [85, 78, 92, 74, 95]

df_scores = pd.DataFrame({"Subject": subjects, "Score": scores})
print(df_scores)

plt.figure(figsize=(8, 5))
ax = sns.barplot(x="Subject", y="Score", hue="Subject", data=df_scores,
                  palette="viridis", legend=False)

for i, v in enumerate(df_scores["Score"]):
    ax.text(i, v + 1, str(v), ha="center", fontweight="bold")

plt.title("Subject-wise Scores")
plt.xlabel("Subject")
plt.ylabel("Score")
plt.grid(axis="y", linestyle="--", alpha=0.7)
plt.show()
```

```python
#Q3

import numpy as np
import matplotlib.pyplot as plt

roll_no = 1024170323
np.random.seed(roll_no)

data = np.random.randn(50)
cumulative = np.cumsum(data)
noise = np.random.randn(50)

fig, axs = plt.subplots(2, 2, figsize=(12, 8))

axs[0, 0].plot(cumulative, color="green")
axs[0, 0].set_title("Cumulative Sum")
axs[0, 0].set_xlabel("Index")
axs[0, 0].set_ylabel("Cumulative Value")
axs[0, 0].grid(True)

axs[0, 1].scatter(range(50), data + noise, color="purple")
axs[0, 1].set_title("Scatter Plot with Random Noise")
axs[0, 1].set_xlabel("Index")
axs[0, 1].set_ylabel("Value")
axs[0, 1].grid(True)

axs[1, 0].plot(data, color="orange", marker="o", markersize=3)
axs[1, 0].set_title("Original Random Data (randn)")
axs[1, 0].set_xlabel("Index")
axs[1, 0].set_ylabel("Value")
axs[1, 0].grid(True)

axs[1, 1].hist(data, bins=10, color="skyblue", edgecolor="black")
axs[1, 1].set_title("Histogram of Random Data")
axs[1, 1].set_xlabel("Value")
axs[1, 1].set_ylabel("Frequency")
axs[1, 1].grid(True)

plt.tight_layout()
plt.show()
```

```python
#Q4

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

sales_df = pd.read_csv("company_sales_data.csv")
print(sales_df)
```

```python
#Q4.1 - Total profit of all months (line plot)

plt.figure(figsize=(9, 5))
sns.lineplot(x="month_number", y="total_profit", data=sales_df, marker="o", color="green")
plt.title("Total Profit per Month")
plt.xlabel("Month")
plt.ylabel("Total Profit")
plt.grid(True)
plt.show()
```

```python
#Q4.2 - All product sales (multiline plot)

products = ["facecream", "facewash", "toothpaste", "bathingsoap", "shampoo", "moisturizer"]

plt.figure(figsize=(10, 6))
for product in products:
    sns.lineplot(x="month_number", y=product, data=sales_df, marker="o", label=product)

plt.title("Product-wise Sales per Month")
plt.xlabel("Month")
plt.ylabel("Units Sold")
plt.legend()
plt.grid(True)
plt.show()
```

```python
#Q4.3 - Bar chart for all features/attributes

feature_totals = sales_df[products].sum()

plt.figure(figsize=(9, 5))
sns.barplot(x=feature_totals.index, y=feature_totals.values,
            hue=feature_totals.index, palette="Set2", legend=False)
plt.title("Total Sales by Product (All Months)")
plt.xlabel("Product")
plt.ylabel("Total Units Sold")
plt.xticks(rotation=20)
plt.grid(axis="y", linestyle="--", alpha=0.6)
plt.show()
```
